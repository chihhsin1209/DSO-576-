```python
#3. Load the original source files

reviews = pd.read_json(
    "data/raw/Appliances.jsonl",
    lines=True
)

meta = pd.read_json(
    "data/raw/meta_Appliances.jsonl",
    lines=True
)

print("Review rows:", len(reviews))
print("Metadata rows:", len(meta))
```

```python
# 4. Inspect columns and types

print("REVIEW DATA TYPES")
print(reviews.dtypes)

print("\nMETADATA TYPES")
print(meta.dtypes)
```

```python
print("Missing review ratings:", reviews["rating"].isna().sum())
print("Missing helpful votes:", reviews["helpful_vote"].isna().sum())

print("Missing metadata average ratings:", meta["average_rating"].isna().sum())
print("Missing metadata rating numbers:", meta["rating_number"].isna().sum())
print("Missing metadata prices:", meta["price"].isna().sum())
```

```python
# 5. Standardize text

# 
reviews_before = reviews[["title", "text", "asin", "parent_asin"]].copy()
meta_before = meta[["title", "store", "parent_asin"]].copy()

# 
for col in ["title", "text"]:
    reviews[col] = reviews[col].str.strip()

for col in ["title", "store"]:
    meta[col] = meta[col].str.strip()

# 
review_changed = reviews_before[
    (reviews_before["title"] != reviews["title"]) |
    (reviews_before["text"] != reviews["text"])
].head(5)

meta_changed = meta_before[
    (meta_before["title"] != meta["title"]) |
    (meta_before["store"] != meta["store"])
].head(5)

print("Changed review examples:")
display(pd.DataFrame({
    "before_title": review_changed["title"],
    "after_title": reviews.loc[review_changed.index, "title"],
    "before_text": review_changed["text"],
    "after_text": reviews.loc[review_changed.index, "text"]
}))

print("\nChanged metadata examples:")
display(pd.DataFrame({
    "before_title": meta_changed["title"],
    "after_title": meta.loc[meta_changed.index, "title"],
    "before_store": meta_changed["store"],
    "after_store": meta.loc[meta_changed.index, "store"]
}))

# Confirm IDs were not changed
print("\nReview ASINs changed:",
      (reviews_before["asin"] != reviews["asin"]).sum())
print("Review parent ASINs changed:",
      (reviews_before["parent_asin"] != reviews["parent_asin"]).sum())
print("Metadata parent ASINs changed:",
      (meta_before["parent_asin"] != meta["parent_asin"]).sum())
```

```python
# 6. Investigate duplicates

review_cols = ["rating", "title", "text", "asin", "parent_asin",
               "user_id", "timestamp", "helpful_vote", "verified_purchase"]

meta_cols = ["main_category", "title", "average_rating",
             "rating_number", "price", "store", "parent_asin"]

print("Duplicate review rows:",
      reviews.duplicated(subset=review_cols).sum())

print("Duplicate metadata rows:",
      meta.duplicated(subset=meta_cols).sum())

# Check repeated IDs
print("Repeated review ASINs:",
      reviews["asin"].duplicated().sum())

print("Repeated review parent ASINs:",
      reviews["parent_asin"].duplicated().sum())

print("Repeated metadata parent ASINs:",
      meta["parent_asin"].duplicated().sum())
```

```python
duplicates = reviews[
    reviews.duplicated(subset=review_cols, keep=False)
]

duplicates[review_cols].sort_values(
    ["parent_asin", "user_id"]
).head(10)
```

```python
# Remove duplicate review rows
before_rows = len(reviews)

reviews = reviews.drop_duplicates(
    subset=review_cols,
    keep="first"
).copy()

after_rows = len(reviews)

print("Review rows before:", before_rows)
print("Review rows after:", after_rows)
print("Rows removed:", before_rows - after_rows)

# Check again
print("Duplicate review rows remaining:",
      reviews.duplicated(subset=review_cols).sum())
```

```python
# 7. Handle missing data

review_cols = ["rating", "title", "text", "parent_asin"]
meta_cols = [
    "title", "features", "description",
    "price", "details", "parent_asin"
]

print("REVIEW DATA - missing values")
print(reviews[review_cols].isna().sum())

print("\nMETADATA - missing values")
print(meta[meta_cols].isna().sum())
```

```python
print("Missing values after cleaning:")
print(meta[meta_cols].isna().sum())
```

```python
# 8. Check suspicious values

print("REVIEW DATA")

print("Ratings outside 1-5:",
      ((reviews["rating"] < 1) | (reviews["rating"] > 5)).sum())

print("Negative helpful votes:",
      (reviews["helpful_vote"] < 0).sum())

print("Earliest review date:",
      reviews["timestamp"].min())

print("Latest review date:",
      reviews["timestamp"].max())


print("\nMETADATA")

print("Average ratings outside 1-5:",
      ((meta["average_rating"] < 1) | (meta["average_rating"] > 5)).sum())

print("Negative rating numbers:",
      (meta["rating_number"] < 0).sum())

print("Prices <= 0:",
      (meta["price"].dropna() <= 0).sum())
```

```python
# 9. Verify merges

print("Review rows before merge:", len(reviews))
print("Metadata rows before merge:", len(meta))

review_keys = set(reviews["parent_asin"])
meta_keys = set(meta["parent_asin"])

unmatched_review_keys = review_keys - meta_keys
unmatched_meta_keys = meta_keys - review_keys

print("Unmatched review parent_asin:", len(unmatched_review_keys))
print("Unmatched metadata parent_asin:", len(unmatched_meta_keys))


merged = reviews.merge(
    meta,
    on="parent_asin",
    how="left",
    suffixes=("_review", "_meta"),
    validate="many_to_one"
)

print("Rows after merge:", len(merged))
print("Unexpected extra rows:", len(merged) - len(reviews))
```

```python
# 10. Reconcile changes

starting_review_rows = 2128605
final_review_rows = len(reviews)

starting_meta_rows = 94327
final_meta_rows = len(meta)

print("Starting review rows:", starting_review_rows)
print("Final review rows:", final_review_rows)
print("Review rows removed:", starting_review_rows - final_review_rows)

print("\nStarting metadata rows:", starting_meta_rows)
print("Final metadata rows:", final_meta_rows)
print("Metadata rows removed:", starting_meta_rows - final_meta_rows)

print("\nMerged rows:", len(merged))
print("Difference between final reviews and merged rows:",
      len(merged) - final_review_rows)
```

```python
# 11. Check calculations

print("Total metadata rows:", len(meta))
print("Non-missing price records:", meta["price"].notna().sum())
print("Missing price records:", meta["price"].isna().sum())

print("Average price:", meta["price"].mean())

print("\nMissing review parent_asin:",
      reviews["parent_asin"].isna().sum())

print("Missing metadata parent_asin:",
      meta["parent_asin"].isna().sum())
```

```python
# 12. Verify individual records

print("1. Normal review")
display(reviews.iloc[[0]][["title", "text", "parent_asin"]])

print("\n2. Review after text cleaning")
display(reviews.iloc[[2429]][["title", "text"]])

print("\n3. Metadata after text cleaning")
display(meta.iloc[[47]][["title", "parent_asin"]])

print("\n4. Metadata record with missing price")
display(meta[meta["price"].isna()].head(1)[["title", "price", "parent_asin"]])

print("\n5. Check product ID")
display(meta.iloc[[0]][["title", "parent_asin"]])

# 12. Five record checks

checks = pd.DataFrame({
    "record": [
        "Review title - row 2429",
        "Metadata title - row 47",
        "Missing price - row 0",
        "Review parent_asin - row 0",
        "Metadata parent_asin - row 0"
    ],

    "original": [
        repr(reviews_before.loc[2429, "title"]),
        repr(meta_before.loc[47, "title"]),
        meta.loc[0, "price"],
        reviews_before.loc[0, "parent_asin"],
        meta_before.loc[0, "parent_asin"]
    ],

    "expected": [
        repr(reviews_before.loc[2429, "title"].strip()),
        repr(meta_before.loc[47, "title"].strip()),
        "NaN (keep missing)",
        reviews_before.loc[0, "parent_asin"],
        meta_before.loc[0, "parent_asin"]
    ],

    "actual": [
        repr(reviews.loc[2429, "title"]),
        repr(meta.loc[47, "title"]),
        meta.loc[0, "price"],
        reviews.loc[0, "parent_asin"],
        meta.loc[0, "parent_asin"]
    ]
})

checks["match"] = [
    reviews.loc[2429, "title"] == reviews_before.loc[2429, "title"].strip(),
    meta.loc[47, "title"] == meta_before.loc[47, "title"].strip(),
    pd.isna(meta.loc[0, "price"]),
    reviews.loc[0, "parent_asin"] == reviews_before.loc[0, "parent_asin"],
    meta.loc[0, "parent_asin"] == meta_before.loc[0, "parent_asin"]
]

checks
```
