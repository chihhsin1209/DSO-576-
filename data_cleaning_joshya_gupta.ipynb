# Amazon Appliances Review Data Cleaning

## Project purpose
This project supports Amazon product-listing quality decisions in the Appliances category. The goal is to identify review records that may mention missing or unclear listing information, which can help prioritize product listings for improvement.

## Data source and scope
Source: McAuley Lab Amazon Reviews 2023 — Appliances review data.

To keep this assignment reproducible and manageable, I downloaded the first 300 review records from the official Appliances review file. The cleaning script saves the downloaded raw sample locally, cleans it, and creates a 50-row cleaned sample.

## Cleaning decisions

| Rule | Reason | Check |
|---|---|---|
| Standardize whitespace in review titles and text | Prevents inconsistent spacing from affecting analysis | Review text is trimmed and blank text becomes missing |
| Convert ratings and helpful-vote counts to numbers | Ensures numeric fields can be analyzed correctly | Invalid ratings and negative helpful votes are changed to missing |
| Convert timestamps to UTC datetimes | Makes dates consistent and usable for analysis | Invalid timestamps become missing |
| Remove duplicate reviews using user ID, product ID, timestamp, and review text | Prevents the same review from being counted more than once | The script reports how many duplicates were removed |
| Create a hashed review ID | Provides a stable, privacy-conscious identifier for each review | Each cleaned row has a `review_uid` |
| Flag possible listing-information issues with review keywords | Creates a starting point for investigating unclear or missing listing details | `issue_mentioned` is True when review text mentions terms such as “missing,” “description,” “dimensions,” “warranty,” or “compatibility” |

## Results from this run
- Reviews loaded: 300
- Exact duplicates removed: 0
- Invalid ratings changed to missing: 0
- Negative helpful votes changed to missing: 0
- Reviews missing text: 0
- Reviews flagged as possible listing-information issues: 12
- Cleaned sample rows saved: 50

## Important limitation
The keyword flag is only a screening tool. A review that mentions “description” or “warranty” does not automatically prove that listing information is missing. In a later project phase, I would review the flagged records alongside the corresponding product metadata using `parent_asin`.
